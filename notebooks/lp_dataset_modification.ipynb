```python
import pandas as pd
```

```python
df = pd.read_csv('/home/islam_tb/Documents/AlgerPlate/data/project-2-at-2026-04-21-15-24-06e8bde7.csv')
df.head()
```

```python
def chg_name(name):
    return name[(name.find('-')+1):]

df['file_name'] = df['img_name'].apply(chg_name)
df.head()
```

```python
df_copy = df.copy()
df_copy.head()
```

```python
df_copy = df_copy.drop(columns=['annotation_id', 'annotator', 'created_at', 'id', 'lead_time', 'updated_at'])
df_copy.head()
```

```python
df_copy = df_copy.drop(columns=['img_name'])
df_copy.head()
```

```python
df_copy = df_copy.drop_duplicates(subset=['file_name'], keep='first')
df_copy.head()
```

```python
df_copy.info()
```

```python
df_copy['text'] = df_copy['text'].astype(str)
```

```python
analysis = pd.DataFrame({
    'Length': df_copy['text'].str.len(),
    'Year_Code': df_copy['text'].str[-4:-2],
    'Wilaya_Code': df_copy['text'].str[-2:]
})

# Example: Get the top 5 most frequent Wilayas
print(analysis['Wilaya_Code'].value_counts().head(5))
```

```python
def reg_year(y):
    if int(y)>26:
        return '19'+y
    else:
        return '20'+y
    
analysis['Year_Code'] = analysis['Year_Code'].apply(reg_year)
analysis
```

```python
# Showing how many plates of each length belong to each Year
ct = pd.crosstab(analysis['Year_Code'], analysis['Length'])
print(ct)
```
