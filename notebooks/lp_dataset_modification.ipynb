```python
analysis = pd.DataFrame({
    'Length': df_copy['text'].str.len(),
    'Year_Code': df_copy['text'].str[-4:-2],
    'Wilaya_Code': df_copy['text'].str[-2:]
})

# Example: Get the top 5 most frequent Wilayas
print(analysis['Wilaya_Code'].value_counts().head(5))
```

```python
def reg_year(y):
    if int(y)>26:
        return '19'+y
    else:
        return '20'+y
    
analysis['Year_Code'] = analysis['Year_Code'].apply(reg_year)
analysis
```

```python
# Showing how many plates of each length belong to each Year
ct = pd.crosstab(analysis['Year_Code'], analysis['Length'])
print(ct)
```

```python
df_copy = df_copy[df_copy['text'].str.len()<12]
df_copy.head(1)
```

```python
df_copy = df_copy.drop(columns=["annotation_id", "annotator", "created_at", "id", "lead_time", "updated_at"])
df_copy.head(1)
```

```python
def get_suffix_after_hyphen(text):
    # split(separator, maxsplit)
    # This splits the string at the first '-' and takes the second part [1]
    parts = text.split('-', 1)
    
    # Check if a hyphen actually exists to avoid IndexErrors
    return parts[1] if len(parts) > 1 else text

df_copy["file_name"] = df_copy["img_name"].apply(get_suffix_after_hyphen)
df_copy.head()
```

```python
df_copy = df_copy.drop(columns=["img_name"])
df_copy.head()
```
